## Die Rechnungen der Lektionen, in Python

Diese Zellen wiederholen die Rechnungen der Lektionen 2.1 bis 2.5. Führen Sie sie aus und vergleichen Sie mit dem Buch.

```python
# Lektion 2.1 — die Regel der Agentur auf G angewandt, Term für Term (Gleichung 2.1)
part_surface = 3 * 62        # k€/m² × m² = k€
part_distance = -8 * 2.5     # k€/km × km = k€
part_fixe = 45               # k€
prix_G = part_surface + part_distance + part_fixe
print(part_surface, part_distance, part_fixe, '→', prix_G, 'k€; Abweichung mit Vorzeichen:', 209 - prix_G, 'k€')
```

```python
# Lektion 2.2 — Vektoren, Skalarprodukt (Gleichung 2.2) und Vorhersage (Gleichung 2.3)
x_G = np.array([62, 2.5])    # (Fläche in m², Entfernung in km)
w = np.array([3, -8])        # (k€/m², k€/km), gleiche Reihenfolge wie x
b = 45                       # k€
print('Dimension von x_G:', x_G.shape, '  w · x_G =', np.dot(w, x_G), '  ŷ =', np.dot(w, x_G) + b)
print('Gewichte in falscher Reihenfolge:', np.dot(w[::-1], x_G) + b, 'k€ (absurd)')
```

```python
# Lektion 2.3 — alle Vorhersagen auf einmal (Gleichung 2.5)
X = logements[['surface_m2', 'distance_km']].to_numpy(dtype=float)
y = logements['prix_kEUR'].to_numpy(dtype=float)
y_chapeau = X @ w + b
print('X :', X.shape, ' w :', w.shape, ' ŷ :', y_chapeau.shape)
print('ŷ =', y_chapeau)
print('Abweichungen mit Vorzeichen y − ŷ =', y - y_chapeau)
print('Mittelwert der absoluten Abweichungen, 7 Wohnungen:', round(np.mean(np.abs(y - y_chapeau)), 2), 'k€')
print('ebenso, A bis F:', round(np.mean(np.abs(y - y_chapeau)[:6]), 2), 'k€; Gerade vom Kapitelanfang, A bis F:',
      round(np.mean(np.abs(y - (3 * X[:, 0] + 30))[:6]), 2), 'k€')
```

```python
# Lektion 2.4 — Indikatorspalten (Gleichung 2.6), Gare als Referenz
indicatrices = pd.get_dummies(est['quartier'], drop_first=True).astype(int)   # Spalten Parc, Rocade
X_est = np.column_stack([est['surface_m2'], est['distance_km'], indicatrices['Parc'], indicatrices['Rocade']])
w_est = np.array([3, -8, 12, -10])
y_est = est['prix_kEUR'].to_numpy(dtype=float)
print(indicatrices.columns.tolist(), ' X :', X_est.shape)
print('mit dem Viertel:', X_est @ w_est + 45)
print('ohne das Viertel:', X_est[:, :2] @ w_est[:2] + 45)
print('Mittelwert der absoluten Abweichungen: ohne', np.mean(np.abs(y_est - (X_est[:, :2] @ w_est[:2] + 45))),
      ' mit', np.mean(np.abs(y_est - (X_est @ w_est + 45))))
```

```python
# Lektion 2.5 — dieselbe Vorhersage auf zwei Skalen (Gleichung 2.7), Division durch n
X5 = t2_sud[['surface_m2', 'distance_km']].to_numpy(dtype=float)
mu, sigma = X5.mean(axis=0), X5.std(axis=0, ddof=0)
w, b = np.array([3, -8]), 45
w_std, b_std = w * sigma, b + np.sum(w * mu)
Z = (X5 - mu) / sigma
print('Mittelwerte:', mu, ' Standardabweichungen:', sigma)
print('standardisierte Gewichte:', w_std, ' Achsenabschnitt:', b_std)
print('ursprüngliche Einheiten:', X5 @ w + b)
print('standardisierte Skala:', Z @ w_std + b_std)
```
