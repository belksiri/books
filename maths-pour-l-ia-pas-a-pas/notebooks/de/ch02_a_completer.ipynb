## Die Rechnungen der Lektionen, in Python

Diese Zellen wiederholen die Rechnungen der Lektionen 2.1 bis 2.5. Führen Sie sie aus und vergleichen Sie mit dem Buch.

```python
# Lektion 2.1 — die Regel der Agentur auf G angewandt, Term für Term (Gleichung 2.1)
area_part = 3 * 62        # k€/m² × m² = k€
distance_part = -8 * 2.5     # k€/km × km = k€
fixed_part = 45               # k€
price_G = area_part + distance_part + fixed_part
print(area_part, distance_part, fixed_part, '→', price_G, 'k€; Abweichung mit Vorzeichen:', 209 - price_G, 'k€')
```

```python
# Lektion 2.2 — Vektoren, Skalarprodukt (Gleichung 2.2) und Vorhersage (Gleichung 2.3)
x_G = np.array([62, 2.5])    # (Fläche in m², Entfernung in km)
w = np.array([3, -8])        # (k€/m², k€/km), gleiche Reihenfolge wie x
b = 45                       # k€
print('Dimension von x_G:', x_G.shape, '  w · x_G =', np.dot(w, x_G), '  ŷ =', np.dot(w, x_G) + b)
print('Gewichte in falscher Reihenfolge:', np.dot(w[::-1], x_G) + b, 'k€ (absurd)')
```

```python
# Lektion 2.3 — alle Vorhersagen auf einmal (Gleichung 2.5)
X = homes[['floor_area', 'distance']].to_numpy(dtype=float)
y = homes['price'].to_numpy(dtype=float)
y_hat = X @ w + b
print('X :', X.shape, ' w :', w.shape, ' ŷ :', y_hat.shape)
print('ŷ =', y_hat)
print('Abweichungen mit Vorzeichen y − ŷ =', y - y_hat)
print('Mittelwert der absoluten Abweichungen, 7 Wohnungen:', round(np.mean(np.abs(y - y_hat)), 2), 'k€')
print('ebenso, A bis F:', round(np.mean(np.abs(y - y_hat)[:6]), 2), 'k€; Gerade vom Kapitelanfang, A bis F:',
      round(np.mean(np.abs(y - (3 * X[:, 0] + 30))[:6]), 2), 'k€')
```

```python
# Lektion 2.4 — Indikatorspalten (Gleichung 2.6), Gare als Referenz
dummies = pd.get_dummies(east['district'], drop_first=True).astype(int)   # Spalten Parc, Rocade
X_east = np.column_stack([east['floor_area'], east['distance'], dummies['Parc'], dummies['Rocade']])
w_east = np.array([3, -8, 12, -10])
y_east = east['price'].to_numpy(dtype=float)
print(dummies.columns.tolist(), ' X:', X_east.shape)
print('mit dem Viertel:', X_east @ w_east + 45)
print('ohne das Viertel:', X_east[:, :2] @ w_east[:2] + 45)
print('mittlere absolute Abweichung: ohne', np.mean(np.abs(y_east - (X_east[:, :2] @ w_east[:2] + 45))),
      ' mit', np.mean(np.abs(y_east - (X_east @ w_east + 45))))
```

```python
# Lektion 2.5 — dieselbe Vorhersage auf zwei Skalen (Gleichung 2.7), Division durch n
X5 = t2_south[['floor_area', 'distance']].to_numpy(dtype=float)
mu, sigma = X5.mean(axis=0), X5.std(axis=0, ddof=0)
w, b = np.array([3, -8]), 45
w_std, b_std = w * sigma, b + np.sum(w * mu)
Z = (X5 - mu) / sigma
print('Mittelwerte:', mu, ' Standardabweichungen:', sigma)
print('standardisierte Gewichte:', w_std, ' Achsenabschnitt:', b_std)
print('ursprüngliche Einheiten:', X5 @ w + b)
print('standardisierte Skala:', Z @ w_std + b_std)
```

```python
# Zwei Versionen derselben Regel: Entfernung in km, dann in Metern
X_km = t2_south[['floor_area', 'distance']].to_numpy(dtype=float)
X_m = X_km * np.array([1, 1000])
w_km, w_m, b = np.array([3, -8]), np.array([3, -0.008]), 45
print('Vorhersagen (km):', X_km @ w_km + b)
print('Vorhersagen (m):  ', X_m @ w_m + b)
results = {}
for name, Xv, wv in [('km', X_km, w_km), ('m', X_m, w_m)]:
    mu, sigma = Xv.mean(axis=0), Xv.std(axis=0, ddof=0)
    results[name] = standardized_weights(wv, b, mu, sigma)
    print(name, ': Mittelwert Entfernung =', mu[1], ', Standardabweichung =', sigma[1], ', standardisierte Gewichte =', results[name])
```
