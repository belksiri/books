```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Farben des Buchs: eine Farbe pro Datensatz
LOT_E, LOT_V, LOT_T = '#2e7d32', '#b84d00', '#1565c0'   # Training (grün), Validierung (dunkelorange), Test (blau)
SPAM, LEGITIME = '#d62728', '#b9b6ae'
VERSION_A, VERSION_B, ECART, SEUIL = '#757575', '#00897b', '#c2185b', '#8e24aa'   # A/B-Test: A grau, B petrol


def sigmoide(z):
    return 1 / (1 + np.exp(-np.asarray(z, dtype=float)))


def perte_moyenne(etiquettes, X, w, b):
    """Mittlerer logarithmischer Verlust (Gleichungen 5.6 und 5.7)."""
    p = np.clip(sigmoide(X @ np.asarray(w, dtype=float) + b), 1e-12, 1 - 1e-12)
    y = np.asarray(etiquettes, dtype=float)
    return float(np.mean(-(y * np.log(p) + (1 - y) * np.log(1 - p))))


NOMS_CARAC = ['liens', 'mots_suspects', 'longueur', 'majuscules', 'heure', 'pieces_jointes', 'taille', 'destinataires']


def fabriquer_messages(graine=77):
    """Fiktive Nachrichten (Aufgabe 6.4 und Lektion 6.4): 40 Gruppen, Spam-Kampagnen (gerade Gruppen) oder
    legitime Absender (ungerade Gruppen). Die Nachrichten einer Gruppe sind fast identisch. Nur die ersten
    beiden Merkmale hängen vom Label ab; die acht Merkmale haben sehr unterschiedliche Einheiten."""
    rng = np.random.default_rng(graine)
    unites = np.array([2, 1.5, 30, 0.2, 5, 1, 50, 3])
    decalages = np.array([3, 2, 200, 1, 12, 0, 500, 10])
    lignes = []
    for g in range(40):
        spam = g % 2 == 0
        taille = rng.integers(4, 15)
        base = rng.normal(0, 1, 8)
        base[:2] += 1.0 if spam else -1.0
        for _ in range(taille):
            lignes.append([g, int(spam)] + list((base + rng.normal(0, 0.3, 8)) * unites + decalages))
    return pd.DataFrame(lignes, columns=['groupe', 'spam'] + NOMS_CARAC)


def ajuster(X, y, lam, l1=False, pas=0.1, n_pas=3000):
    """Gradientenabstieg auf dem bestraften Verlust: L2 (Gleichung 6.2) oder L1 (Gleichung 6.3). b wird nicht bestraft."""
    w, b = np.zeros(X.shape[1]), 0.0
    for _ in range(n_pas):
        ecart = sigmoide(X @ w + b) - y
        gw, gb = X.T @ ecart / len(y), ecart.mean()
        if l1:   # Gradientenschritt, dann jedes Gewicht um höchstens pas × λ an 0 heranrücken (ohne Vorzeichenwechsel)
            w = w - pas * gw
            w = np.sign(w) * np.maximum(np.abs(w) - pas * lam, 0)
        else:    # die Ableitung von λ × Summe der w² ist 2λw
            w = w - pas * (gw + 2 * lam * w)
        b -= pas * gb
    return w, b
```

```python
# Lektion 6.2 — ein Datenleck: Recall und Kosten pro Nachricht der beiden Aufteilungen
for _, l in fuite.iterrows():
    manques = l['manques_copies'] + l['manques_nouveaux']
    nouveaux = l['spams'] - l['copies_vues']
    print(f"{l['decoupage']:>12} : Recall = {(l['spams'] - manques) / l['spams']:.2f}, "
          f"Kosten pro Nachricht = {(manques + 3 * l['fp']) / l['messages']:.3f}, "
          f"Anteil verpasster neuer Spams = {l['manques_nouveaux']} / {nouveaux} = {l['manques_nouveaux'] / nouveaux:.2f}")

# Die Kampagnen aufteilen, nicht die Nachrichten: kleine Demonstration mit 40 Kampagnen unterschiedlicher Größe
rng = np.random.default_rng(6)
campagne = np.repeat(np.arange(40), rng.integers(5, 40, size=40))          # Kampagnennummer jedes Spams
au_hasard = rng.random(campagne.size) < 0.25                              # 25 % der NACHRICHTEN in den Test
camp_test = rng.choice(40, size=10, replace=False)                         # 25 % der KAMPAGNEN in den Test
par_campagne = np.isin(campagne, camp_test)
for nom, test_ in (('zufällig', au_hasard), ('je Kampagne', par_campagne)):
    partagees = len(set(campagne[test_]) & set(campagne[~test_]))
    print(f'{nom:>12} : {test_.sum()} Spams im Test, {partagees} Kampagnen sowohl im Test als auch anderswo vorhanden')
```

```python
# Lektion 6.3 — Überanpassung: Filter A und B auf den Trainingsdaten (zwölf Nachrichten) und den Validierungsdaten (sechs)
X12, y12 = douze[['liens', 'mots_suspects']].to_numpy(float), (douze['etiquette'] == 'spam').to_numpy(int)
X6, y6 = six[['liens', 'mots_suspects']].to_numpy(float), (six['etiquette'] == 'spam').to_numpy(int)
filtres = {'A': ((1, 0.5), -4), 'B': ((-0.54, 4.39), -9.17)}
for nom, (w, b) in filtres.items():
    p6 = sigmoide(X6 @ np.array(w) + b)
    pertes6 = -(y6 * np.log(p6) + (1 - y6) * np.log(1 - p6))
    print(f'Filter {nom}: Training {perte_moyenne(y12, X12, w, b):.2f}, Validierung {perte_moyenne(y6, X6, w, b):.2f} '
          f'| Verluste der sechs: {np.round(pertes6, 2)}')

# Bestrafter L2-Verlust (Gleichung 6.2), λ = 0,05; die Konstante b wird nicht bestraft
lam = 0.05
for nom, (w, b) in filtres.items():
    pen = lam * np.sum(np.square(w))
    print(f'Filter {nom}: Summe der quadrierten Gewichte = {np.sum(np.square(w)):.4f}, Strafterm = {pen:.4f}, '
          f'bestrafter Verlust = {perte_moyenne(y12, X12, w, b) + pen:.2f}')

# Der Strafterm hängt von den Einheiten ab: Links in Zehnern gezählt, gleiche Vorhersagen, anderer Strafterm
X12_diz = X12 / np.array([10, 1])
w_diz = (10, 0.5)
print('gleiche Wahrscheinlichkeiten:', np.allclose(sigmoide(X12_diz @ np.array(w_diz) - 4), sigmoide(X12 @ np.array([1, 0.5]) - 4)),
      f'| Strafterm in Einern: {lam * (1 + 0.25):.4f}; in Zehnern: {lam * (100 + 0.25):.4f}')
```

```python
# Lektion 6.4 — Standardisierung in jeder Runde: Mittelwert und Standardabweichung der Links auf den BEIDEN ANDEREN Folds
for k in range(3):
    autres = plis[plis['pli'] != k + 1]
    n = autres['messages'].sum()
    m = (autres['messages'] * autres['moyenne_liens']).sum() / n
    # Standardabweichung der Vereinigung zweier Folds (Division durch n): Mittelwert der Quadrate minus Quadrat des Mittelwerts
    v = (autres['messages'] * (autres['ecart_type_liens'] ** 2 + autres['moyenne_liens'] ** 2)).sum() / n - m ** 2
    print(f'Runde {k + 1}: Validierung = Fold {k + 1}, Training = {n} Nachrichten, Mittelwert {m:.2f}, Standardabweichung {np.sqrt(v):.2f}')
print(f'Nachricht mit 6 Links aus Fold 3, in Runde 3: (6 - 2,25) / 1,76 = {(6 - 2.25) / 1.76:.2f}')

# Kreuzvalidierungswert (Gleichung 6.4) der beiden Werte von λ
for col, lam in (('perte_l001', 0.01), ('perte_l01', 0.1)):
    print(f'λ = {lam}: Verluste der Folds {list(plis[col])}, KV = {plis[col].mean():.3f}')
```

```python
# Lektion 6.4 (Schluss) — L1 und L2 auf fiktiven Nachrichten mit acht Merkmalen (dieselben wie in Aufgabe 6.4)
msgs = fabriquer_messages()
X8 = msgs[NOMS_CARAC].to_numpy(float)
X8 = (X8 - X8.mean(0)) / X8.std(0)        # hier Standardisierung über alle Nachrichten: nur zur Demonstration
y8 = msgs['spam'].to_numpy(float)
w_l2, _ = ajuster(X8, y8, 0.1)
w_l1, _ = ajuster(X8, y8, 0.1, l1=True)
print('Gewichte L2:', np.round(w_l2, 2))
print('Gewichte L1:', np.round(w_l1, 2))
print(f'Nullgewichte: L2 {int(np.sum(np.abs(w_l2) < 1e-9))}, L1 {int(np.sum(np.abs(w_l1) < 1e-9))} '
      '(nur die ersten beiden Merkmale hängen mit dem Label zusammen)')
```

```python
# Lektionen 6.5 und 6.6 — der A/B-Test der Buchhandlung
a, b_ = librairie.iloc[0], librairie.iloc[1]
pA, pB = a['commandes'] / a['clients'], b_['commandes'] / b_['clients']
ecart = pB - pA
print(f'Rate A = {100 * pA:.1f} %, Rate B = {100 * pB:.1f} %, Differenz = {100 * ecart:.1f} Punkte, '
      f'relative Veränderung = {100 * ecart / pA:.0f} %')
esA, esB = np.sqrt(pA * (1 - pA) / a['clients']), np.sqrt(pB * (1 - pB) / b_['clients'])
es = np.sqrt(esA ** 2 + esB ** 2)                          # Gleichung 6.6: man addiert die QUADRATE
bas, haut = ecart - 1.96 * es, ecart + 1.96 * es           # Gleichung 6.7
print(f'SF von A = {100 * esA:.2f} Punkt, SF von B = {100 * esB:.2f} Punkt, SF der Differenz = {100 * es:.2f} Punkt')
print(f'95 %-Konfidenzintervall: von {100 * bas:.1f} bis {100 * haut:.1f} Punkten')
print(f'für 40 000 Abonnenten: {40000 * ecart:.0f} Bestellungen mehr (von {40000 * bas:.0f} bis {40000 * haut:.0f})')

plt.figure(figsize=(6, 1.8))
plt.axvline(0, color='black', linewidth=2.5, label='0: kein Effekt')
plt.axvline(1, color=SEUIL, linestyle='--', label='Rentabilitätsschwelle: 1 Punkt')
plt.plot([100 * bas, 100 * haut], [0, 0], color=ECART, linewidth=4, label='95 %-Intervall')
plt.scatter([100 * ecart], [0], marker='D', s=70, color=ECART, zorder=3)
plt.yticks([]); plt.xlim(-1, 5); plt.xlabel('Effekt der Version B (Prozentpunkte)')
plt.legend(fontsize=7, loc='upper right'); plt.show()

# Lektion 6.6 — die Kosten pro Nachricht im Test (Lektion 6.1) sind ein Mittelwert: gleiche Überlegung
couts = np.array([1] * 20 + [3] * 4 + [0] * 226)            # 20 verpasste Spams, 4 Fehlalarme, 226 ohne Kosten
es_c = couts.std() / np.sqrt(len(couts))                     # Standardabweichung (Division durch n) / Wurzel aus n
print(f'Kosten pro Nachricht = {couts.mean():.3f}, Standardabweichung = {couts.std():.3f}, SF = {es_c:.3f}, '
      f'Intervall: von {couts.mean() - 1.96 * es_c:.2f} bis {couts.mean() + 1.96 * es_c:.2f}')
```
