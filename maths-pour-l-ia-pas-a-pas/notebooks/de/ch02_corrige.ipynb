## Aufgabe 2.5 — Vorhersagen, dann prüfen: die Einheit wechseln

*Daten D2.5. Begriffe: Skalenwechsel (Lektion 2.5, Gleichung 2.7).* Notieren Sie Ihre Vorhersagen zu a) und b), bevor Sie die Zellen ausführen.

**Erwartete Vorhersagen.** a) Das Gewicht der Entfernung wird zu −0,008 k€ pro Meter. Der Achsenabschnitt bleibt 45 k€, und die fünf Vorhersagen ändern sich nicht. b) Der Mittelwert geht von 2 km auf 2000 m, die Standardabweichung von 1 km auf 1000 m. Das standardisierte Gewicht bleibt $w'_2 = -0{,}008 \times 1\,000 = -8$ k€ pro Standardabweichung: Es hängt nicht von der gewählten Einheit ab.

```python
def standardized_weights(w, b, mu, sigma):
    w_std = w * sigma              # k€ pro Standardabweichung
    b_std = b + np.sum(w * mu)     # vorhergesagter Preis der mittleren Wohnung
    return w_std, b_std
```

```python
# Zwei Versionen derselben Regel: Entfernung in km, dann in Metern
X_km = t2_south[['floor_area', 'distance']].to_numpy(dtype=float)
X_m = X_km * np.array([1, 1000])
w_km, w_m, b = np.array([3, -8]), np.array([3, -0.008]), 45
print('Vorhersagen (km):', X_km @ w_km + b)
print('Vorhersagen (m):  ', X_m @ w_m + b)
results = {}
for name, Xv, wv in [('km', X_km, w_km), ('m', X_m, w_m)]:
    mu, sigma = Xv.mean(axis=0), Xv.std(axis=0, ddof=0)
    results[name] = standardized_weights(wv, b, mu, sigma)
    print(name, ': Mittelwert Entfernung =', mu[1], ', Standardabweichung =', sigma[1], ', standardisierte Gewichte =', results[name])
```

```python
# Prüfung
if results['km'][0] is None:
    print('Ergänzen Sie die Funktion standardized_weights.')
else:
    ok = all(np.allclose(results[k][0], [12, -8]) and np.isclose(results[k][1], 164) for k in results)
    print('standardisierte Gewichte in km und in m identisch:', 'richtig' if ok else 'bitte prüfen')
```

```python
if results['km'][0] is not None:
    fig, axes = plt.subplots(1, 2, figsize=(8, 2.6))
    for ax, title, values in [
            (axes[0], 'Rohgewicht des 2. Terms', {'km': w_km[1], 'm': w_m[1]}),
            (axes[1], 'Standardisiertes Gewicht des 2. Terms', {k: results[k][0][1] for k in results})]:
        ax.bar(list(values), list(values.values()), color=[BLUE, ORANGE])
        ax.axhline(0, color='black', linewidth=0.8)
        ax.set_title(title, fontsize=10)
    plt.tight_layout()
    plt.show()
```

### Lösung zu Aufgabe 2.5

**a)** Ein Meter ist 1/1000 km: Das Gewicht wird zu −8 / 1000 = **−0,008 k€/m**. Da jede Entfernung mit 1000 multipliziert wird, bleibt jedes Produkt $w_2 x_2$ gleich: Der Achsenabschnitt (45 k€) und die fünf Vorhersagen (146, 149, 165, 163, 197 k€) **ändern sich nicht**.

**b)** Mittelwert 2000 m, Standardabweichung 1000 m. Mit Gleichung 2.7: $w'_2 = -0{,}008 \times 1\,000 = -8$ und $b' = 45 + 3 \times 45 - 0{,}008 \times 2\,000 = 45 + 135 - 16 = 164$ k€, **genau** wie in km.

**c)** Im linken Diagramm geht das Rohgewicht von −8 auf −0,008: Seine Größe hängt von der Einheit ab und erlaubt keinen Vergleich. Im rechten stimmen die standardisierten Gewichte überein: Ein Anstieg der Entfernung um eine Standardabweichung entspricht immer −8 k€ in der Vorhersage, bei fester Fläche. Diese Lesart beschreibt die Regel auf diesen Daten. Sie misst weder eine allgemeine Wichtigkeit noch einen kausalen Effekt.
