## Los cálculos de las lecciones, en Python

Estas celdas rehacen los cálculos de las lecciones 2.1 a 2.5. Ejecútalas y compáralas con el libro.

```python
# Lección 2.1 — la regla de la agencia aplicada a G, término a término (ecuación 2.1)
part_surface = 3 * 62        # k€/m² × m² = k€
part_distance = -8 * 2.5     # k€/km × km = k€
part_fixe = 45               # k€
prix_G = part_surface + part_distance + part_fixe
print(part_surface, part_distance, part_fixe, '→', prix_G, 'k€; desviación con signo:', 209 - prix_G, 'k€')
```

```python
# Lección 2.2 — vectores, producto escalar (ecuación 2.2) y predicción (ecuación 2.3)
x_G = np.array([62, 2.5])    # (superficie en m², distancia en km)
w = np.array([3, -8])        # (k€/m², k€/km), mismo orden que x
b = 45                       # k€
print('dimensión de x_G:', x_G.shape, '  w · x_G =', np.dot(w, x_G), '  ŷ =', np.dot(w, x_G) + b)
print('pesos en el orden equivocado:', np.dot(w[::-1], x_G) + b, 'k€ (absurdo)')
```

```python
# Lección 2.3 — todas las predicciones de una vez (ecuación 2.5)
X = logements[['surface_m2', 'distance_km']].to_numpy(dtype=float)
y = logements['prix_kEUR'].to_numpy(dtype=float)
y_chapeau = X @ w + b
print('X :', X.shape, ' w :', w.shape, ' ŷ :', y_chapeau.shape)
print('ŷ =', y_chapeau)
print('desviaciones con signo y − ŷ =', y - y_chapeau)
print('media de las desviaciones en valor absoluto, 7 viviendas:', round(np.mean(np.abs(y - y_chapeau)), 2), 'k€')
print('ídem, A a F:', round(np.mean(np.abs(y - y_chapeau)[:6]), 2), 'k€; recta de la apertura del capítulo, A a F:',
      round(np.mean(np.abs(y - (3 * X[:, 0] + 30))[:6]), 2), 'k€')
```

```python
# Lección 2.4 — columnas indicadoras (ecuación 2.6), Gare como referencia
indicatrices = pd.get_dummies(est['quartier'], drop_first=True).astype(int)   # columnas Parc, Rocade
X_est = np.column_stack([est['surface_m2'], est['distance_km'], indicatrices['Parc'], indicatrices['Rocade']])
w_est = np.array([3, -8, 12, -10])
y_est = est['prix_kEUR'].to_numpy(dtype=float)
print(indicatrices.columns.tolist(), ' X :', X_est.shape)
print('con el barrio:', X_est @ w_est + 45)
print('sin el barrio:', X_est[:, :2] @ w_est[:2] + 45)
print('media de las desviaciones en valor absoluto: sin', np.mean(np.abs(y_est - (X_est[:, :2] @ w_est[:2] + 45))),
      ' con', np.mean(np.abs(y_est - (X_est @ w_est + 45))))
```

```python
# Lección 2.5 — la misma predicción en dos escalas (ecuación 2.7), división por n
X5 = t2_sud[['surface_m2', 'distance_km']].to_numpy(dtype=float)
mu, sigma = X5.mean(axis=0), X5.std(axis=0, ddof=0)
w, b = np.array([3, -8]), 45
w_std, b_std = w * sigma, b + np.sum(w * mu)
Z = (X5 - mu) / sigma
print('medias:', mu, ' desviaciones típicas:', sigma)
print('pesos estandarizados:', w_std, ' ordenada:', b_std)
print('unidades originales:', X5 @ w + b)
print('escala estandarizada:', Z @ w_std + b_std)
```
